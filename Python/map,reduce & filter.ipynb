### **Map, Reduce & Filter**

#### **Area of a circle**

```python
import math

def area(r):
    return math.pi*(r**2)

radii = [1,2,3,4,5]
areas = []
for r in radii:
    a = area(r)
    areas.append(a)
```

```python
areas
```

```python
map(area,radii)
```

```python
list(map(area,radii))
```

#### **Convert Celcius scale to Farheneit**

```python
temps = [("Mumbai",35), ("Berlin",12),("Tokyo",25),("Hong kong",31),("Sydney",301)]
```

```python
type(temps)
```

```python
cel_to_f = lambda data: (data[0], (9/5)*[1]+32)
```
